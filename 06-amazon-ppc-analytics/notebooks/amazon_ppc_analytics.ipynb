# Amazon PPC Analytics

Campaign efficiency, spend allocation, and weekly trend analysis.

**Dataset:** campaign, keyword-type, and weekly PPC metrics supplied with this project.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / 'data'
campaign = pd.read_excel(DATA_DIR / 'campaign_performance.xls')
spend = pd.read_excel(DATA_DIR / 'spend_allocation.xls')
weekly = pd.read_excel(DATA_DIR / 'weekly_trends.xls', parse_dates=['week'])
```

## 1. Campaign Performance

Compare spend, revenue, ROAS, ACOS, CTR, CVR, and CPA at campaign level.

```python
campaign[['campaign','match_type','keyword_type','spend','revenue','orders','ROAS','ACOS','CTR','CVR','CPA']].sort_values('ROAS', ascending=False)
```

## 2. Spend Allocation

Compare spend share and return across keyword types.

```python
spend.sort_values('ROAS', ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.bar(spend['keyword_type'], spend['spend_pct'])
ax.set_title('Advertising Spend Allocation')
ax.set_ylabel('Spend Share (%)')
plt.tight_layout()
plt.show()
```

## 3. Weekly Trends

Track advertising efficiency over time.

```python
weekly[['week','spend','revenue','ROAS','ACOS','CTR']].sort_values('week')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(weekly['week'], weekly['ROAS'], marker='o', label='ROAS')
ax.set_title('Weekly ROAS Trend')
ax.set_ylabel('ROAS')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Management Takeaways

Use campaign-level and weekly metrics to investigate efficiency, spend allocation, and changes in advertising performance. These observations are descriptive and do not establish causality.
