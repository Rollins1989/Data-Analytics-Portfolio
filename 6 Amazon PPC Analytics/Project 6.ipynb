# Amazon PPC Analytics

Campaign efficiency, spend allocation, and weekly trend analysis.

**Dataset:** campaign, keyword-type, and weekly PPC metrics supplied with this project.


```python
import pandas as pd
import matplotlib.pyplot as plt

campaign = pd.read_csv('campaign_performance.xls')
spend = pd.read_csv('spend_allocation.xls')
weekly = pd.read_csv('weekly_trends.xls', parse_dates=['week'])
```

## 1. Campaign Performance

Compare spend, revenue, ROAS, ACOS, CTR, CVR, and CPA at campaign level.

```python
campaign[['campaign','match_type','keyword_type','spend','revenue','orders','ROAS','ACOS','CTR','CVR','CPA']].sort_values('ROAS', ascending=False)
```

## 2. Spend Allocation

Brand keywords generate the strongest return in the supplied allocation table, while competitor keywords consume a relatively large share of spend for a much smaller share of revenue.

```python
spend.sort_values('ROAS', ascending=False)
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.bar(spend['keyword_type'], spend['spend_pct'])
ax.set_title('Advertising Spend Allocation')
ax.set_ylabel('Spend Share (%)')
plt.tight_layout()
plt.show()
```

## 3. Weekly Trends

Track whether advertising efficiency is improving or deteriorating over time.

```python
weekly[['week','spend','revenue','ROAS','ACOS','CTR']].sort_values('week')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.plot(weekly['week'], weekly['ROAS'], marker='o', label='ROAS')
ax.set_title('Weekly ROAS Trend')
ax.set_ylabel('ROAS')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 4. Management Takeaways

1. Protect efficient Brand and high-ROAS campaigns from budget constraints.
2. Review high-ACOS competitor and broad-match campaigns for bid, targeting, and search-term efficiency.
3. Use weekly ROAS and ACOS trends to detect performance deterioration before reallocating budget.
4. These observations are descriptive; they do not establish that a specific bidding change caused performance changes.
