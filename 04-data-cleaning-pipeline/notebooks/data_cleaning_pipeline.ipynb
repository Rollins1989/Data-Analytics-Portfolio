# Data Cleaning Pipeline

An auditable, reproducible workflow for detecting and repairing common tabular data-quality issues.

**Data provenance:** synthetic employee data generated locally for demonstration.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_DIR = Path.cwd().parent
sys.path.insert(0, str(PROJECT_DIR / 'src'))

from data_cleaning_pipeline import generate_messy_data, clean_data, save_quality_visualization
```

```python
# Generate deterministic messy data.
raw = generate_messy_data(n=5000, seed=77)
print(f'Raw shape: {raw.shape}')
print(f'Initial missing values: {int(raw.isna().sum().sum())}')
raw.head()
```

```python
# Clean and audit the dataset.
cleaned, report = clean_data(raw)
report_dict = report.__dict__
pd.Series(report_dict, name='value')
```

```python
# Review the cleaned output and verify key quality conditions.
print(f'Cleaned shape: {cleaned.shape}')
print(f'Missing values after cleaning: {int(cleaned.isna().sum().sum())}')
print(f'Duplicate rows after cleaning: {int(cleaned.duplicated().sum())}')
cleaned.head()
```

```python
# Save the same deliverables produced by the reusable script.
OUTPUT_DIR = PROJECT_DIR / 'outputs'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
cleaned.to_csv(OUTPUT_DIR / 'cleaned_data.csv', index=False)
pd.Series(report_dict).to_json(OUTPUT_DIR / 'pipeline_run_report.json', indent=2)
save_quality_visualization(report)
print('Saved: cleaned_data.csv, pipeline_run_report.json, data_quality_output.png')
```
